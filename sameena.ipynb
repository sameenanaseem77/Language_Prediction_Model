```python
from sklearn.feature_extraction.text import CountVectorizer #example

#Create a CountVectorizer obj
vectorizer =   CountVectorizer()

#Sample  text data
data= ["love data science", "love machine learning"]

#Fit and transform the data
vectorised_data = vectorizer.fit_transform(data)

#Get the vocabulary (unique words)
print(vectorizer.get_feature_names_out())

#Convert the results to an array #(bag of words) is present in sklearn which contain all the unique words of the data
print (vectorised_data.toarray()) #toarray is converting words into number
#bag of words only care about words not sequence or alphabets
```
